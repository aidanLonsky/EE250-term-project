## Step 1 - Combine and Filter

- load and merge all csv data files (for 2024, months Jan - Mar)
- Slice dataframe to view only flights departing from airport of interest (I've chosen **SFO**)

```python
# Import data into dataframe and inspect
from pathlib import Path
import pandas as pd

files = sorted(Path("../data").glob("*.csv"))

if not files:
    raise FileNotFoundError("No CSV files found in data/")

df = pd.concat((pd.read_csv(file) for file in files), ignore_index=True)

core_df = df[df["ORIGIN"]=="SFO"]
```

## Step 2 - Drop incomplete flights (dropped or diverted)

```python
drop_mask = (df["CANCELLED"] == 1.0) | (df["DIVERTED"] == 1.0)

rows_dropped = drop_mask.sum()
core_df = core_df.loc[~drop_mask].copy()
row_remaining = len(core_df)

print(f"Rows dropped: {rows_dropped:,}")
print(f"Rows remaining: {row_remaining:,}")
core_df
```

## Step 3 - Create seed for random sample seeding going forwards

```python
seed = 6847
sample = core_df.sample(frac=0.1, random_state=seed)
sample
```
